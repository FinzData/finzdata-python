# 13F holdings in Python

Find a manager, read its 13F book with weights, list a stock's holders and track ownership flows.

Set `FINZDATA_API_KEY` first (free key: https://api.finzdata.com/). Cells marked as needing a look-back bundle return HTTP 403 on free keys.

## Finding a Manager by Name

```python
import finzdata as yf
# Needs a look-back bundle or the Institutional plan (free keys get HTTP 403)
c = yf.Client()
managers = c.managers(q="berkshire")
print(managers[['cik', 'manager', 'filing_date', 'positions', 'value_usd']].head())
```

## Retrieving a Manager's Full 13F Book

```python
import finzdata as yf
# Needs a look-back bundle or the Institutional plan (free keys get HTTP 403)
c = yf.Client()
berkshire_holdings = c.manager_holdings(cik="1067983", period="2026Q2")
print(berkshire_holdings[['ticker', 'issuer', 'shares', 'value_usd', 'weight']].head())
```

## Finding All Holders of a Specific Stock

```python
import finzdata as yf
# Needs a look-back bundle or the Institutional plan (free keys get HTTP 403)
c = yf.Client()
apple_holders = c.holders(ticker="AAPL", period="2026Q2")
print(apple_holders[['manager', 'shares', 'value_usd', 'filing_date']].head())
```

## Tracking Ownership Flows by Quarter

```python
import finzdata as yf
c = yf.Client()
aapl_ownership = c.institutional_ownership(ticker="AAPL")
print(aapl_ownership[['period_end', 'holders', 'shares', 'shares_change', 'new_positions', 'closed_positions']])
```
