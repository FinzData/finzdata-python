# Futures: continuous series, single contracts and the roll calendar

[FinzData futures](https://finzdata.com/futures-data): 139 markets (CME, CBOT, NYMEX, COMEX, ICE, Eurex, Cboe), 17,625 individual contracts since 2007, back-adjusted and raw continuous series, and the roll calendar. Bars: 1, 5, 15, 30-minute, hourly and daily.

A free key returns the last 30 days of daily bars for all 139 markets. One market's full history, every bar size, is $49 ([shop](https://finzdata.com/shop)).

```python
import finzdata as fz
c = fz.get_client()

es = c.futures("ES", continuous=True)  # back-adjusted continuous E-mini S&P 500
es.tail()
```

## One contract at its own traded prices

```python
front = es["contract"].iloc[-1]
print("front contract:", front)
c.futures("ES", contract=front).tail()
```

## The roll calendar

One row per front-month contract: first and last front-month day, expiry, the next contract and the roll date used to build the continuous series. Use it to stitch contracts yourself or to check a backtest rolled when you think it did.

```python
c.futures_rolls("ES").tail()
```
